## PyPSA Example

```python
import pandas as pd
import pypsa
```

```python
n = pypsa.Network()
```

```python
n.add("Bus", "gen_bus", carrier="transmission")
n.add("Bus", "load_bus")
n.add("Load", "load_1", bus="load_bus", p_set=500)
n.add(
    "Link", "transmission", bus0="gen_bus", bus1="load_bus", efficiency=0.93, p_nom=1000
)
```

```python
n.add(
    "Generator",
    "coal",
    bus="gen_bus",
    p_nom=300,
    p_nom_min=50,
    marginal_cost=3,
)
n.add(
    "Generator",
    "gas",
    bus="gen_bus",
    p_nom=400,
    p_nom_min=100,
    marginal_cost=4,
)
n.generators
```

```python
n.links
```

```python
n.optimize()
```

```python
n.buses_t.marginal_price
```

```python
n.generators_t.p
```

```python
n.model.to_file("test2.lp")
```

```python
n.plot()
```
