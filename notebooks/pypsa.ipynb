## PyPSA Example

```python
import pypsa
```

```python
n = pypsa.Network()
```

```python
n.add("Bus", "bus_1")
n.add("Load", "load_1", bus="bus_1", p_set=500)
```

```python
n.add(
    "Generator",
    "coal",
    bus="bus_1",
    p_nom=300,
    p_nom_min=50,
    marginal_cost=3,
    committable=True,
)
n.add(
    "Generator",
    "gas",
    bus="bus_1",
    p_nom=400,
    p_nom_min=100,
    marginal_cost=4,
)
n.generators
```

```python
n.optimize()
```

```python
n.generators_t.p
```

```python
n.buses_t.marginal_price
```

```python
n.lines_t.p1
```

```python
n.model.to_file("test2.lp")
```
